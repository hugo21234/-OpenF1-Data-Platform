```sql
SELECT DISTINCT
    source_table_full_name AS origem,
    target_table_full_name AS destino,
    1 AS peso

FROM system.access.table_lineage

WHERE
    source_table_full_name IS NOT NULL
    AND target_table_full_name IS NOT NULL

    AND source_table_full_name LIKE 'f1_plataform_data.%'
    AND target_table_full_name LIKE 'f1_plataform_data.%';
```
