```sql
CREATE OR REPLACE TABLE pits AS
SELECT
    to_timestamp(date) AS date,
    session_key,
    pit_duration,
    meeting_key,
    TRY_CAST(driver_number AS INT) as driver_number,
    stop_duration,
    lane_duration,
    TRY_CAST(lap_number AS INT) as lap_number
FROM f1_plataform_data.bronze.pits
```
