```sql
CREATE OR REPLACE TABLE stints AS 
SELECT     
    meeting_key,     
    session_key,     
    CAST(stint_number AS INT) AS stint_number,     
    CAST(driver_number AS INT) AS driver_number,     
    CAST(lap_start AS INT) AS lap_start,     
    CAST(lap_end AS INT) AS lap_end,     
    TRIM(compound) AS compound,     
    CAST(tyre_age_at_start AS INT) AS tyre_age_at_start 
FROM f1_plataform_data.bronze.stints
```
