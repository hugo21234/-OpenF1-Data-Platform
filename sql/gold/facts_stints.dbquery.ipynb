```sql
CREATE OR REPLACE TABLE fact_stints AS
    SELECT
        uuid() AS stint_id,
        d.driver_id,
        s.compound,
        s.lap_start,
        s.lap_end,
        s.stint_number,
        s.tyre_age_at_start,
        s.meeting_key,
        s.session_key


FROM f1_plataform_data.silver.stints AS s
JOIN dim_drivers AS d
ON s.driver_number = d.driver_number
```
