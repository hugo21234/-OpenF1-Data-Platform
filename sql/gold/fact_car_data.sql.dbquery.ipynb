```sql
CREATE OR REPLACE TABLE facts_car_data AS
SELECT
    uuid() as car_data_id,
    d.driver_id,
    c.brake_boolean,
    c.date,
    c.drs,
    c.meeting_key,
    c.session_key,
    c.n_gear,
    c.rpm,
    c.speed,
    c.throttle

FROM f1_plataform_data.silver.car_data AS c
JOIN dim_drivers AS d
ON c.driver_number = d.driver_number
```
