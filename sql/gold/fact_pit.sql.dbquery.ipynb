```sql
CREATE OR REPLACE TABLE fact_pit AS
SELECT 
    uuid() AS pit_id,
    p.date,
    d.driver_id,
    p.lane_duration,
    p.pit_duration,
    p.stop_duration,
    p.lap_number,
    p.meeting_key,
    p.session_key

    
FROM f1_plataform_data.silver.pits AS p
JOIN dim_drivers AS d
ON  p.driver_number = d.driver_number
WHERE d.is_current = true
```
