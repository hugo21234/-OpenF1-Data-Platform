```sql
CREATE OR REPLACE TABLE fact_laps AS
SELECT
    uuid() as lap_id,
    d.driver_id,
    l.date_start,
    l.duration_sector_1,
    l.duration_sector_2,
    l.duration_sector_3,
    l.i1_speed,
    l.i2_speed,
    l.is_pit_out_lap,
    l.lap_number,
    l.meeting_key,
    l.session_key,
    l.st_speed

FROM f1_plataform_data.silver.laps as l
JOIN dim_drivers as d
ON l.driver_number = d.driver_number

```
